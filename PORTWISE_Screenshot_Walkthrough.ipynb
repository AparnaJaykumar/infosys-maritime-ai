# PORTWISE Milestone 1 Screenshot Walkthrough

## User authentication interface

This notebook brings the ten supplied screenshots and their explanations into one file. It documents login, signup, the two password-recovery entry pages, and separate administrator login in light and dark themes.

All images are embedded in the notebook; no separate image folder or code execution is required to read it. Screenshot content is unchanged. Images are encoded as JPEG at their original dimensions to keep this file compact.

These captures document the visible pages. They do not by themselves establish successful authentication, OTP delivery, password recovery, or dashboard access.

## 1. Login page in light mode



The sign-in page provides fields for a username or email and a password. It also offers navigation to account creation, password recovery, and separate administrator sign-in. The fields are empty in this capture; it documents the interface before authentication.


## 2. Signup page in light mode



The signup form shows username, email address, password, confirm password, security question, and security answer. The password guidance appears beneath the fields. This capture shows the registration form; it does not show a completed signup or email activation.


## 3. Email OTP recovery in light mode



Email OTP is selected as the recovery method. The user can enter a registered email address and request a verification code. This screenshot shows the request stage, before the code-entry form or delivery confirmation.


## 4. Security question recovery in light mode



Security question is selected as the recovery method. The first step asks for the registered username so that the saved question can be retrieved. The saved question, answer verification, and new-password form are not shown here.


## 5. Administrator login in light mode



The separate administration page asks for an administrator username and password. Its sign-in button leads to the admin dashboard after successful verification. This is the login page, not the registered-user directory.


## 6. Login page in dark mode



The login page is shown with the dark maritime background and darker form controls. The same username or email and password fields and navigation options remain available. This documents the alternate theme.


## 7. Signup page in dark mode



The dark version retains all six signup fields, including password confirmation and security question selection. Password guidance remains visible below the form fields.


## 8. Email OTP recovery in dark mode



The dark recovery page displays the registered-email field and the Send verification code action with Email OTP selected. As in the light capture, this is the initial request stage rather than evidence of email delivery or successful verification.


## 9. Security question recovery in dark mode



The dark recovery page asks for a registered username with Security question selected. The Find security question button begins this route; the later answer and password-reset stages are not pictured.


## 10. Administrator login in dark mode



The administrator sign-in form is shown in the dark theme. It remains separate from the regular user login and includes a return-to-sign-in action. A screenshot taken after a successful admin login is still needed to document the user directory.
