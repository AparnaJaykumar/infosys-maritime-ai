## Member 1 — RAG builder workflow

The supplied RAG builder is intended to develop and persist a freight/logistics knowledge base. Its documented flow includes: **source collection → URL/content deduplication → HTML/PDF extraction → cleaning → collection gate for 150+ unique sources → chunk generation and deduplication → persistent export → sentence embeddings and FAISS → BM25 retrieval → knowledge-answer tests with numbered citations**.

Coverage targets include Incoterms, HS codes, customs procedures, shipping documents, container/cargo types, freight and surcharges, demurrage/detention, ports/congestion, weather/route risks, insurance and trade finance. The reference notebook names `FREIGHT_SEEDS`, stores output under a `rag_data` directory, and includes a source-count gate rather than treating the 150-source requirement as optional.

**What to validate in Colab:** at least 150 successfully collected *unique* sources; source/PDF statistics; actual `chunks.jsonl`, `sources.csv`, relevant database and index files saved in Drive; FAISS and BM25 loading from the integrated app; 10 knowledge tests; short retrieved answers with verifiable citation markers such as `[1]` and `[2]`. The builder's code supports the workflow, but the generated corpus/index files were not supplied in this ZIP.

## Five specialist simulators and Digital Twin

**Route simulator:** compare route alternatives under a canal/strait closure or fuel-price shock, reporting expected duration and cost across three routes.

**Quote simulator:** vary margin, fuel and exchange rate to assess price bands and win likelihood.

**Carrier simulator:** factor carrier punctuality and capacity shortages into ranking during disruption.

**Weather simulator:** vary intensity and duration, observing delay estimates and associated demurrage cost.

**Customs simulator:** vary inspection rate and documentation problems, observing clearance delay and daily holding expense.

For every simulator, capture assumptions, fixed random seed, input data provenance, 10th–90th percentile interval, and repeatability. Simulations must **not mutate operational live tables**. The screenshot evidence includes Digital Twin scenario controls and several agent views; it does not independently certify that all five upgraded simulations meet the full specification. Test them explicitly.

## Data quality, auditability and provenance

The platform mixes several categories of information: seeded demonstration datasets, database-backed operational records, possible external weather observations, retrieved freight-reference documents and model-generated natural-language explanations. A production-quality demo should label what is **live**, **simulated**, **derived**, or **retrieved**.

Recommended acceptance evidence: an import that correctly rejects bad rows without changing valid data; user activity/audit entries after login and edits; a quote requiring approval; an agent result grounded in current table values; a citation pointing to an actual knowledge document; a permission-denied server response; and a repeatable Digital Twin result when given the same seed.

## Troubleshooting and careful claims

- **No GPU / slow startup:** select a T4 GPU runtime and rerun from the beginning.
- **Module not found:** run the cells that create `/content/app/` and write modules before running the application server.
- **`rag_data` empty:** run the standalone RAG builder in the correct Drive directory, then ensure the integrated app reads that same path.
- **No knowledge citations:** check sources, chunks, embedding/index files and retrieval configuration before claiming RAG works.
- **ngrok tunnel conflict:** close a previous tunnel or use the Colab proxy fallback; never share tokens.
- **Package conflicts:** use the version constraints in the notebook; a previous Colab installation may need runtime restart.
- **Inconsistent app roles:** inspect `ROLE_MENU`/`ROLE_CAPS` and server-side access checks; do not equate hidden links with security.
- **No proof of answer-rate improvement:** do not report 50/50 until the required questions have actually been retested and counters/screenshots saved.
- **Notebook outputs cleared:** this is intentional to make source submission cleaner; rerun to recreate the live demonstration.
