### Application at a glance

`Visitor → Customer registration/OTP or assigned-role sign-in → JWT identity → FastAPI permission checks → permitted pages and actions → SQLite records → SQL Copilot / freight agents`

A customer can self-register **only as Customer**. Admins provision or change internal roles. The sign-in role selector verifies an existing assigned role; it **cannot** grant privileges.


## How a freight quote works

1. An authorized user enters origin, destination, cargo, weight, HS code and customer information.
2. The existing pipeline evaluates route, carrier, weather, pricing, customs and approval conditions.
3. The system displays the quote price, margin, estimated transit and status; documentation requirements can place a quote on hold.
4. Users can inspect carrier rankings, the route map, weather indicators, price components and customs records, subject to their role.
5. The SQL Copilot can retrieve selected facts from the database using supported question patterns.

**Data-source note:** Open-Meteo weather/marine information is presented as live/cached where configured. Vessel positions in the demonstration are **simulated** and port congestion is **model-predicted**; injected storms are test scenarios. The public homepage's example quote is illustrative.


## Authentication, security and profile

- **Customer registration:** Username, email, strong password, confirmation, security question and answer. Public registration does not choose an elevated role.
- **Password rules:** At least eight characters with uppercase, lowercase, a digit and special character. Passwords are stored as **bcrypt hashes**, not plaintext.
- **Email verification:** A six-digit code sent by email; the demonstrated flow uses a five-minute expiry and rate-limited resends.
- **Password recovery:** Email OTP or verification of the account's security answer, with request throttling.
- **Account lock:** Five incorrect sign-in attempts lock the account; only an administrator can unlock it. The login screen shows **“Account locked”** rather than “Session expired.” A locked user's question may be displayed, but the recovery flow cannot bypass the administrative lock.
- **JWT sessions:** The backend verifies the signed token and the account's current lock/role status for protected requests.
- **My Profile:** Shows username, email, role, joined date and last active session. Users can update the username and password after entering the current password.

**Privacy:** Personal email addresses and sensitive form entries visible in the original captures have been redacted in the screenshots included here. Keep real account data and email/SMTP secrets out of public repositories.


## Five-role access control

| Role | Copilot | Route | Quotes | Carriers | Weather | Margin | Customs | Run agents | Users/admin |
|---|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| **Admin** | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ |
| **Operations Manager** | ✓ | ✓ | — | ✓ | ✓ | — | ✓ | ✓ | — |
| **Freight Broker** | ✓ | — | ✓ | ✓ | — | ✓ | — | — | — |
| **Auditor** (read-only) | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ | — | — |
| **Customer** (own records only) | ✓ | — | **Own** | — | — | — | — | — | — |

**Enforcement is on the FastAPI backend.** Removing a navigation item alone is not security: a request for an unauthorized endpoint must return **HTTP 403 Forbidden**. The Auditor can review the modules but may not create quotes or trigger agents; Customers can access only records linked to their own customer ID.

### Admin Dashboard

Admin users can provision accounts, assign roles, link customers to freight records, lock/unlock or remove users where allowed, view statistics and inspect the latest audit events. Safeguards prevent an administrator from locking/deleting their own account or removing the final Admin. No passwords or password hashes are shown in the user table.


## SQL Copilot: supported questions

The reference Copilot is **SQL-only**; it is **not an LLM, RAG system or unrestricted chatbot**. The enhanced Copilot matches natural-language-like freight requests against **18 whitelisted SQL query definitions** and retrieves actual records. It can display its SQL query and the number of rows returned. Question wording/spelling tolerance is limited to the implemented matching rules.

**Examples of the 18 query categories:**

| # | Example question | Response from database |
|---:|---|---|
| 1 | Which carrier is cheapest to Rotterdam? | Average freight quote prices by matching destination |
| 2 | Which shipments are on customs hold? | Held shipments and identifying shipment fields |
| 3 | Show quotes below 10% margin | Quotes below the given percentage threshold |
| 4 | Which ports have the highest congestion? | Highest congestion indices and dwell times |
| 5 | Show storms near the Red Sea | Matching weather advisories / severe wind records |
| 6 | Which shipments have high delay risk? | Risk-flagged shipments |
| 7 | How many shipments are delayed? | Delayed/held status counts |
| 8 | Which carrier is most reliable? | Carrier on-time-performance ranking |
| 9 | What are my active shipments? | Customer-linked shipments |
| 10 | Show my recent quotes | Customer-linked quote list |
| 11 | Who are the top customers? | Quote-spend leaders, authorized roles only |
| 12 | Which products have high customs duty? | HS-chapter customs rules and duties |
| 13 | Show active vessels at sea | Vessel progress and modeled positions |
| 14 | Show quotes by carrier | Quote counts, prices and margins by carrier |
| 15 | Show shipments from Mumbai | Shipment counts by origin port |
| 16 | Which ports have longest dwell time? | Average port dwell-day ranking |
| 17 | Show severe weather risks | Port weather and winds |
| 18 | What is the fleet health score? | Most recent health/decision record |

**Example outcomes captured on the website:** A greeting such as **“hi”** returns a welcome message and example prompts. **“Which shipments are on customs hold?”** displayed a shipment table with **10 rows** in an earlier captured state. **“Which carrier is cheapest to Rotterdam?”** returned **0 matching rows** in another screenshot; the Copilot reported no records instead of inventing a price. These are **snapshot results**, not guaranteed live values.

**Voice:** The browser microphone can transcribe a spoken question; a *Read answer* control can speak the returned response. Browser support, permission and HTTPS/localhost constraints apply. Voice does not remove the SQL question restrictions.

**RBAC:** Customer Copilot queries must be scoped to their own account. Unsupported requests get a polite fallback with valid example prompts. Weather/location filters and live data should be verified against actual source records before relying on an answer.

## How to run

### Google Colab (primary demonstration method)

1. Open **`FreightQuote_AI_Milestone2_Final_Deployment.ipynb`** and select a Python runtime.
2. In Colab's **Secrets** panel, set a strong `JWT_SECRET` (32+ characters) and `ADMIN_PASSWORD` (a strong unique 12–72-byte password). Add `ADMIN_EMAIL` where used by your configuration. For actual verification emails, also set `EMAIL_ADDRESS` and `EMAIL_PASSWORD` (the SMTP application password).
3. Run the notebook from top to bottom. The repaired Colab startup uses the active Python environment instead of creating a failing `ensurepip` virtual environment.
4. Open the displayed Colab application link. If a public tunnel is needed, set `NGROK_AUTHTOKEN` securely and use the notebook's supported ngrok fallback.
5. Sign in as an administrator using the assigned **Admin** role, or register a Customer and verify email.

**Colab storage is temporary.** Preserve a database separately if records must survive runtime resets. Never paste passwords, API keys or tokens into committed notebook cells.

### Docker (optional local deployment)

```bash
cp .env.example .env
# Set unique JWT_SECRET, ADMIN_PASSWORD and optional email credentials in .env
docker compose up --build
```

Open `http://localhost:8000` locally; for public deployment, add a secure HTTPS reverse proxy. Use a single configured worker/replica for this SQLite-based demo. See the Google Colab and Docker instructions in this section for details.
